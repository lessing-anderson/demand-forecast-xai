# 2. Data Processing

```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd()))
sys.path.insert(0, str(Path.cwd().parent))
from src.data.loader import load_data
from src.data.data_processing import reduce_mem_usage, process_dim_calendar, process_dim_location, process_dim_prices, process_bridge_snap, process_fact_sales

out_dir = Path('../data/processed')
out_dir.mkdir(parents=True, exist_ok=True)
```

## Load Raw Data

```python
raw_path = '../data/raw'

print("Loading raw M5 data...")
calendar, prices, sales = load_data(raw_path)

print(f"Calendar shape: {calendar.shape}")
print(f"Prices shape: {prices.shape}")
print(f"Sales shape: {sales.shape}")
```

## Create Calendar Dimension

```python
# Create copy df from loaded `calendar` DataFrame and reduce memory usage

df_cal = calendar.copy()

df_cal = reduce_mem_usage(df_cal)

# Process calendar data to create dim_calendar
dim_calendar = process_dim_calendar(df_cal, out_dir)

print(f"\nWritten dim_calendar to {out_dir} (rows={len(dim_calendar)})")
print(f"dim_calendar columns: {dim_calendar.columns.tolist()}")
```

## Create Location Dimension

```python
# Create copy df from loaded `sales` DataFrame and reduce memory usage

df_sales = sales.copy()

df_sales = reduce_mem_usage(df_sales)

# Process sales data to create dim_location
dim_location = process_dim_location(df_sales, out_dir)

print(f"\nWritten dim_location to {out_dir} (rows={len(dim_location)})")
print(f"dim_location columns: {dim_location.columns.tolist()}")
```

## Create Price Dimension

```python
# Create copy df from loaded `prices` DataFrame and reduce memory usage

df_prices = prices.copy()

df_prices = reduce_mem_usage(df_prices)

# Process prices data to create dim_prices
dim_prices = process_dim_prices(df_prices, out_dir)

print(f"\nWritten dim_prices to {out_dir} (rows={len(dim_prices)})")
print(f"dim_prices columns: {dim_prices.columns.tolist()}")
```

## Create SNAP Bridge Factless Fact

```python
# Create copy df from loaded `calendar` DataFrame and reduce memory usage

df_bridge_snap = calendar.copy()

df_bridge_snap = reduce_mem_usage(df_bridge_snap)

# Process calendar data to create bridge_SNAP
bridge_SNAP = process_bridge_snap(df_bridge_snap, out_dir)

print(f"\nWritten bridge_SNAP to {out_dir} (rows={len(bridge_SNAP)})")
print(f"bridge_SNAP columns: {bridge_SNAP.columns.tolist()}")
```

## Create Sales Fact

```python
# Create copy df from loaded `sales` DataFrame and reduce memory usage

df_sales = sales.copy()

df_sales = reduce_mem_usage(df_sales)

# Process sales data to create fact_sales
fact_sales = process_fact_sales(df_sales, out_dir)

print(f"\nWritten fact_sales to {out_dir} (rows={len(fact_sales)})")
print(f"fact_sales columns: {fact_sales.columns.tolist()}")
```
